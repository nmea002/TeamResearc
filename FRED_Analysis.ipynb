# FRED Economic Data Analysis

This notebook provides tools for analyzing Federal Reserve Economic Data (FRED) from CSV files. Use this to load, explore, and visualize economic time series data.

## 1. Import Required Libraries

Import libraries for data handling, visualization, and API access.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')

# Set style for better visualizations
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
```

## 2. Load FRED CSV Data

Load economic time series data from FRED CSV files. You can download CSV files from https://fred.stlouisfed.org

```python
# Function to load FRED CSV files
def load_fred_csv(file_path):
    """
    Load a FRED CSV file into a DataFrame
    
    Args:
        file_path (str): Path to the CSV file
        
    Returns:
        pd.DataFrame: The loaded data
    """
    df = pd.read_csv(file_path)
    print(f"Loaded {file_path}")
    print(f"Shape: {df.shape}")
    return df

# Example: Load a single CSV file
# Uncomment and modify the path to your FRED CSV file
# df = load_fred_csv('path/to/your/fred_data.csv')

# Or load multiple FRED files from a directory
def load_multiple_fred_files(directory):
    """
    Load all CSV files from a directory into a dictionary of DataFrames
    
    Args:
        directory (str): Path to the directory containing CSV files
        
    Returns:
        dict: Dictionary with filenames as keys and DataFrames as values
    """
    dataframes = {}
    if os.path.exists(directory):
        for file in os.listdir(directory):
            if file.endswith('.csv'):
                file_path = os.path.join(directory, file)
                dataframes[file] = pd.read_csv(file_path)
                print(f"Loaded: {file}")
    return dataframes

# Example usage:
# data_dir = 'path/to/your/csv/folder'
# fred_files = load_multiple_fred_files(data_dir)
```

## 3. Parse and Clean Data

Handle date parsing, missing values, and data type conversions for FRED time series data.

```python
def clean_fred_data(df):
    """
    Clean and parse FRED data
    
    Args:
        df (pd.DataFrame): Input DataFrame from FRED CSV
        
    Returns:
        pd.DataFrame: Cleaned DataFrame
    """
    df_clean = df.copy()
    
    # Rename common FRED column names
    if 'DATE' in df_clean.columns:
        df_clean.rename(columns={'DATE': 'date'}, inplace=True)
    
    # Parse date column
    if 'date' in df_clean.columns:
        df_clean['date'] = pd.to_datetime(df_clean['date'])
        df_clean = df_clean.set_index('date')
    
    # Handle missing values
    # Replace '.' or 'NaN' with np.nan
    df_clean = df_clean.replace('.', np.nan)
    df_clean = df_clean.replace('nan', np.nan)
    
    # Convert numeric columns to float
    for col in df_clean.columns:
        try:
            df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')
        except:
            pass
    
    # Display data info
    print("Data Info:")
    print(f"Shape: {df_clean.shape}")
    print(f"Date Range: {df_clean.index.min()} to {df_clean.index.max()}")
    print(f"\nMissing Values:\n{df_clean.isnull().sum()}")
    
    return df_clean

# Example usage:
# df_clean = clean_fred_data(df)
```
