```python
import matplotlib.pyplot as plt

data = {"Male": [1.00, 0.19], "Female" : [0.99, 0.20], 
       "18-30 years old": [0.99, 0.20], "31-40 years old": [1.01, 0.17], "41-50 years old": [0.98, 0.23], "Over 50 years old": [0.99, 0.14], 
       "Married": [1.01, 0.18], "Unmarried": [0.97, 0.20]}

x = list(data.keys())
y = [data[key][0] for key in x]
yerr = [data[key][1] for key in x]
plt.figure(figsize=(10, 6))
plt.bar(x, y, yerr=yerr, capsize=5)
mean = sum(y) / len(y)
plt.axhline(mean, color='red', linestyle='--')
plt.xlabel('Demographic Groups')
plt.ylabel('Mean Portfolio Beta')
plt.title('Mean Portfolio Beta by Demographic Group with Error Bars (Standard Deviation)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
