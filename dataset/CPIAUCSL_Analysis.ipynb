```python
# observation_date is a date string in YYYY-MM-DD format (no time component).
dates = pd.to_datetime(cpiaucsl_data['observation_date'], format='%Y-%m-%d')
cpiaucsl_data['Month'] = dates.dt.month
cpiaucsl_data['Year'] = dates.dt.year

cpiaucsl_data.head()
```
